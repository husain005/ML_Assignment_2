```python
import pandas as pd
import numpy as np
import joblib

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.neighbors import KNeighborsClassifier

from sklearn.naive_bayes import GaussianNB

from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef
)
```

```python
data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = data.target
```

```python
print(X.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
test_df = X_test.copy()

test_df["target"] = y_test

test_df.to_csv(
    "test_data.csv",
    index=False
)

print("CSV created")
```

```python
logistic = LogisticRegression(
    max_iter=5000
)

logistic.fit(
    X_train,
    y_train
)

joblib.dump(
    logistic,
    "logistic.pkl"
)
```

```python
tree = DecisionTreeClassifier()

tree.fit(
    X_train,
    y_train
)

joblib.dump(
    tree,
    "tree.pkl"
)
```

```python
knn = KNeighborsClassifier()

knn.fit(
    X_train,
    y_train
)

joblib.dump(
    knn,
    "knn.pkl"
)
```

```python
nb = GaussianNB()

nb.fit(
    X_train,
    y_train
)

joblib.dump(
    nb,
    "nb.pkl"
)
```

```python
svm = SVC(
    probability=True
)

svm.fit(
    X_train,
    y_train
)

joblib.dump(
    svm,
    "svm.pkl"
)
```

```python
def evaluate(name, model):

    pred = model.predict(X_test)

    prob = model.predict_proba(X_test)[:,1]

    return [
        name,
        accuracy_score(y_test,pred),
        roc_auc_score(y_test,prob),
        precision_score(y_test,pred),
        recall_score(y_test,pred),
        f1_score(y_test,pred),
        matthews_corrcoef(y_test,pred)
    ]
```

```python
results = []

results.append(
    evaluate(
        "Logistic Regression",
        logistic
    )
)

results.append(
    evaluate(
        "Decision Tree",
        tree
    )
)

results.append(
    evaluate(
        "KNN",
        knn
    )
)

results.append(
    evaluate(
        "Naive Bayes",
        nb
    )
)

results.append(
    evaluate(
        "SVM",
        svm
    )
)
```

```python
result_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "AUC",
        "Precision",
        "Recall",
        "F1",
        "MCC"
    ]
)

result_df
```

```python
result_df.sort_values(
    by="Accuracy",
    ascending=False
)
```
